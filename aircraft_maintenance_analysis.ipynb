```python
import pandas as pd

df = pd.read_csv("aircraft_maintenance.csv")

df
```

```python
df.shape
```

```python
len(df)
```

```python
## Question 1: Which defect category occurs most frequently?
df["Defect_Category"].value_counts()
```

```python
## Question 2: What is the average rectification time?
average_hours = df["Rectification_Hours"].mean()

print("Average rectification time:", average_hours, "hours")
```

```python
## Question 3: Which aircraft type has the higher average rectification time?
df.groupby("Aircraft")["Rectification_Hours"].mean()
```

```python
### Finding

#The A321 has a higher average rectification time of 3.1 hours compared with 2.5 hours for the A320 in this sample dataset.
```

```python
## Question 4: How does the average rectification time compare between aircraft types?
average_by_aircraft = df.groupby("Aircraft")["Rectification_Hours"].mean()

average_by_aircraft.plot(kind="bar")
```

```python
average_by_aircraft = df.groupby("Aircraft")["Rectification_Hours"].mean()

average_by_aircraft.plot(
    kind="bar",
    title="Average Rectification Time by Aircraft Type",
    xlabel="Aircraft Type",
    ylabel="Average Rectification Hours"
)
```

```python
### Finding

#The A321 has a higher average rectification time than the A320 in this sample dataset.
```
